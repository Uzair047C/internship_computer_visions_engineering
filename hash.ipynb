```python

import glob, imagehash
from PIL import Image

h = lambda p: imagehash.phash(Image.open(p).convert("L"), hash_size=8)
train = {p: h(p) for p in glob.glob("data_Set\\train\\train\\images*.jpg")}

rows = []
for p in glob.glob("data_Set\\test\\images\\*.jpg"):
    ht = h(p)
    d, q = min((ht - hh, qq) for qq, hh in train.items())
    rows.append((d, p, q))
rows.sort()

print("min Hamming distance:", rows[0][0])
print("test images with a train image at distance <= 10:", sum(r[0] <= 10 for r in rows))
for r in rows[:10]:
    print(r)
```
